```python
import pandas as pd
import matplotlib as plt
import seaborn
import sqlite3

carList_df = pd.read_csv("vehicles_dataset.csv")
recallsList_df = pd.read_csv("RCL_FROM_2020_2024.csv")
#recallsList_df.columns = recallsList_df.columns.str.lower() ##################################################################################
#(Lower casing)
#cars_df["make"] = cars_df["make"].str.lower().str.strip()
#recalls_df["make"] = recalls_df["make"].str.lower().str.strip()
#recalls_df["model"] = recalls_df["model"].str.lower().str.strip()


conn = sqlite3.connect("Joined_Data.db") #create a SQLite database

# Write both Dataframes to seperate tables
carList_df.to_sql("cars", conn, if_exists="replace", index=False)
recallsList_df.to_sql("recalls", conn, if_exists="replace", index=False)

# INNER JOIN on NHTSA ID (can also use LEFT JOIN if you want to keep all cars even without recalls)
query = """
SELECT cars.*, recalls.[NHTSA ID], recalls.summary, recalls.[DOCUMENT NAME]
FROM cars
JOIN recalls
ON cars.[MAKE] = recalls.[MAKE]
"""

joined_df = pd.read_sql_query(query, conn)
print(joined_df.head())

joined_df.to_csv("joined_cars_with_recalls.csv", index=False)


joined_df.shape
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Group by 'make' and count recalls (assuming one row = one recall)
recall_counts = joined_df.groupby('make')['file].count().sort_values(ascending=False)

# Plot
plt.figure(figsize=(12, 6))
plt.bar(recall_counts.index, recall_counts.values)
plt.xlabel('Car Make')
plt.ylabel('Number of Recalls')
plt.title('Number of Recalls per Car Make')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
